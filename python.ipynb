### 1. Dado un número, imprime si es par o impar. (if/else)

```python
num = int(input("Enter a number: "))

if num  % 2 == 0:
    print(f"{num} is an even number.")
else: 
    print(f"{num} isn't a even number.")
```

### 2. Dado un número, imprime si es positivo, negativo o cero. (if/elif/else)

```python
num = int(input("Ingresa un número entero: "))

if num > 0:
    print("Es mayor a 0")
elif num < 0:
    print("Es menor a 0")
else:
    print("Es 0")
```

### 3. Dadas tres calificaciones, imprime si el promedio aprueba (≥6) o no. (comparación)   

```python
cal1 = float(input("Ingresa la primera calificación: "))
cal2 = float(input("Ingresa la segunda calificación: "))
cal3 = float(input("Ingresa la tercer calificación: "))

prom = (cal1 + cal2 + cal3)/(3)

if prom >= 6.0:
    print(f"{prom} Aprueba")
else:
    print(f"{prom} Reprueba")
```
